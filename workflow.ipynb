# Resume Project Workflow

This notebook documents the current beginner-friendly project structure.

- The old optimizer still uses LangGraph in `pipeline.py`.
- The GitHub resume builder now uses clear two-phase Python runner functions in `builder_pipeline.py`.
- Runtime trace is stored as compact `event_trace`, not LLM-generated thought summaries.

## Current File Map

| File | Purpose |
| --- | --- |
| `app.py` | Streamlit entry point and mode selector |
| `builder_app.py` | Streamlit UI for GitHub resume builder |
| `pipeline.py` | Existing resume optimizer workflow |
| `builder_pipeline.py` | GitHub builder workflow and compact events |
| `model_client.py` | Gemini/Groq JSON client and fallback |
| `optimizer_llm.py` | Prompts for optimizer mode |
| `builder_llm.py` | Prompts for builder mode |
| `resume_format.py` | Resume JSON schema, cleanup, rendering |
| `llm.py` | Small compatibility import wrapper |

```python
"""Show the current GitHub builder workflow.

The builder is intentionally simple now.  Streamlit runs phase 1, pauses for
student/user answers, then runs phase 2.
"""

phase_1 = [
    "input",
    "source_check",
    "github_fetch",
    "readme_analysis",
    "linkedin_safe",
    "resume_parse",
    "question_generation",
]

phase_2 = [
    "human_answers",
    "profile_build",
    "project_ranking",
    "resume_write",
    "claim_validation",
    "human_review",
    "export",
]

print("GitHub Resume Builder Workflow")
print("\nPhase 1: collect evidence and ask missing questions")
for step in phase_1:
    print(f"- {step}")

print("\nPhase 2: build, validate, review, export")
for step in phase_2:
    print(f"- {step}")
```

```python
"""Write a simple Mermaid workflow file for project documentation."""

builder_mermaid = """
flowchart TD
    A[input] --> B[source_check]
    B --> C[github_fetch]
    C --> D[readme_analysis]
    D --> E[linkedin_safe]
    E --> F[resume_parse]
    F --> G[question_generation]
    G --> H[human_answers]
    H --> I[profile_build]
    I --> J[project_ranking]
    J --> K[resume_write]
    K --> L[claim_validation]
    L --> M[human_review]
    M --> N[export]
""".strip()

with open("builder_workflow.mmd", "w", encoding="utf-8") as file:
    file.write(builder_mermaid + "\n")

print("Saved builder_workflow.mmd")
```
